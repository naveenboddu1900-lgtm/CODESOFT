```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set()

df = pd.read_csv("comments.csv")

print(df.head())
print(df.columns)

df['date'] = pd.to_datetime(df['posted date'], errors='coerce')
df = df.dropna(subset=['date'])

df['hour'] = df['date'].dt.hour
df['day'] = df['date'].dt.name

# Check for required columns before calculating engagement
required_engagement_cols = ['likes', 'comments', 'followers']
if all(col in df.columns for col in required_engagement_cols):
    for col in required_engagement_cols:
        df[col] = df[col].fillna(df[col].median())

    df['engagement'] = (df['likes'] + df['comments']) / df['followers']

    df.replace([np.inf, -np.inf], np.nan, inplace=True)
    df.dropna(subset=['engagement'], inplace=True)

    hourly = df.groupby('hour')['engagement'].mean()

    plt.figure()
    hourly.plot(marker='o')
    plt.show()

    daily = df.groupby('day')['engagement'].mean()

    plt.figure()
    daily.plot(kind='bar')
    plt.show()

    if 'content_type' in df.columns:
        content = df.groupby('content_type')['engagement'].mean()
        plt.figure()
        content.plot(kind='bar')
        plt.show()

    if 'followers' in df.columns:
        df['followers_diff'] = df['followers'].diff()
        plt.figure()
        df['followers_diff'].plot()
        plt.show()

    pivot = df.pivot_table(values='engagement', index='day', columns='hour', aggfunc='mean')

    plt.figure(figsize=(10,6))
    sns.heatmap(pivot, cmap='coolwarm')
    plt.show()

    print(hourly.idxmax())
    print(daily.idxmax())

    if 'content_type' in df.columns:
        print(content.idxmax())
else:
    print("Warning: Cannot calculate engagement. Missing one or more of 'likes', 'comments', or 'followers' columns in the DataFrame.")
    print("The available columns are:", df.columns.tolist())

# This block is independent of engagement calculation and can run if 'hashtags' column exists
if 'hashtags' in df.columns:
    df['hashtags'] = df['hashtags'].fillna('').astype(str)
    df['hashtags'] = df['hashtags'].str.split(',')
    tags = df['hashtags'].explode()
    top_tags = tags.value_counts().head(10)
    plt.figure()
    top_tags.plot(kind='bar')
    plt.show()
```
